# Implementation of OLS

$$\theta_0=\bar{y}-\theta_1\bar{x}$$

$$\theta_1=\frac{\sum_{i=1}^{N} x^i(y^i-\bar{y})}{\sum_{i=1}^{N} x^i(x^i-\bar{x})}$$

```python
def mean(x):
    sum = 0
    for i in range(len(x)):
        sum = sum + x[i]
        pass
    ave = sum/len(x)
    return ave
```

```python
mean(y)
```

```python
# obtaining theta1:
xMean = mean(x)
yMean = mean(y)
num = 0
den = 0
for i in range(len(x)):
    num = num + x[i]*(y[i]-yMean)
    den = den + x[i]*(x[i]-xMean)
    pass
t1H = num/den
t0H = yMean-t1H*xMean
print(t0H, t1H)
```

```python
def ols(x,y):
    xMean = mean(x)
    yMean = mean(y)
    num = 0
    den = 0
    for i in range(len(x)):
        num = num + x[i]*(y[i]-yMean)
        den = den + x[i]*(x[i]-xMean)
    pass
    t1H = num/den
    t0H = yMean-t1H*xMean
    return t0H, t1H
```

```python
ols(x,y)
```

```python
ols(x,y2)
```

# Comparing data and model performance

```python
# data
plt.plot(x,y2,'.k', label="data")
t0H, t1H = ols(x,y2)
print(t0H, t1H)
yH = t0H+t1H*x
plt.plot(x,yH, ':r', label="Model")
plt.legend()
plt.show()
```
